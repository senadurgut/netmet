```python
def resolution_comparison(file1, file2,
                          tree_name="Events",
                          hwpt_branch="L1EmulEtSum_hwPt",
                          type_branch="L1EmulEtSum_etSumType"):
    """Return flat array of HwPt for etSumType==2."""
    hwpt1 = get_hwpt_met(file1, tree_name, hwpt_branch, type_branch)
    hwpt2 = get_hwpt_met(file2, tree_name, hwpt_branch, type_branch)

    puppi_pt1 = get_puppi_pt(file1, tree_name)
    puppi_pt2 = get_puppi_pt(file2, tree_name)

    diff = hwpt1 - puppi_pt1
    diff2 = hwpt2 - puppi_pt2

    plt.hist(diff, bins=100, alpha=0.5, label='new')
    plt.hist(diff2, bins=100, alpha=0.5, label='old')
    plt.legend()
    plt.show()
```
